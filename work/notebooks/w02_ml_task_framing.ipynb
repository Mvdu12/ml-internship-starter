## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

This lane is a Clustering task. There is no pre-defined outcome to predict —
the goal is to discover natural groupings ("content archetypes") among pages
based on structural properties of the content itself (length, age, update
recency, and type/intent). Clustering fits because I want to reveal hidden
structure in the data, not predict a known label.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

There is no traditional target — this is unsupervised. The proxy is the
combination of structural features that define an archetype: word_count,
char_count, content_age_days, days_since_last_update, content_type, and
main_intent. These describe what a page IS, not an observed outcome. I
deliberately excluded trend_direction and trend_pct since those are the
label source for the rest of the pipeline and must never be used as features,
even here.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

Silhouette score, to confirm the discovered clusters are well-separated and
internally consistent. As a secondary, business-relevant check, I will
compare average word_count and days_since_last_update across clusters — if
the archetypes don't differ meaningfully on these, the grouping isn't
useful for the content team's prioritization.

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

One row = one page. Below is the structural slice relevant to this lane,
after checking missingness (word_count/char_count are blank for content
without a measured length — I check this before deciding how to handle it,
rather than blindly filling with 0).

```python
import pandas as pd

url = 'https://raw.githubusercontent.com/Mvdu12/ml-internship-starter/main/data/raw/content_refresh_anonymized.csv'
df = pd.read_csv(url)

archetype_cols = ["content_type", "word_count", "char_count",
                   "content_age_days", "days_since_last_update", "main_intent"]
lane_df = df[archetype_cols].copy()

print(lane_df.shape[0], "pages | one row = one page")
print("\nMissing word_count by content_type:")
print(df.groupby("content_type")["word_count"].apply(lambda s: s.isna().mean().round(3)))

lane_df.head()
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A hand-written rule (e.g. "word_count > 3000 AND days_since_last_update > 180
= archetype X") forces arbitrary thresholds on several interacting structural
signals at once, across content_type and main_intent categories that behave
differently. With 30,000 pages and multiple continuous + categorical
features, the real groupings are not simple cutoffs — they overlap in ways a
single if/else can't capture. A clustering algorithm finds the natural
groupings from the data itself, instead of me guessing the boundaries.
